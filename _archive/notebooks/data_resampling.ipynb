```python
import numpy as np
import pandas as pd

# Biased row count function
def biased_row_count():
    choices = list(range(162, 171))
    weights = [1, 1, 2, 3, 5, 10, 20, 8, 5]  # Bias toward 168
    return np.random.choice(choices, p=np.array(weights)/sum(weights))

# Find all PPG columns
ppg_cols = [col for col in ppg_bp_data_online_df.columns if col.startswith("PPG_")]

# Randomly select a consecutive block of columns (165–170)
range_len = np.random.randint(165, 171)
start_idx = np.random.randint(0, len(ppg_cols) - range_len + 1)
selected_ppg_cols = ppg_cols[start_idx:start_idx + range_len]

# Resample each selected column with a biased row count (preserve index)
resampled_data = {}
for col in selected_ppg_cols:
    num_rows = biased_row_count()
    resampled_data[col] = ppg_bp_data_online_df[col].iloc[:num_rows]

# Combine into a DataFrame (preserving index for joining)
ppg_resampled_df = pd.DataFrame(resampled_data)

# Now you can join this with demographics using index
# Example: final_df = ppg_resampled_df.join(demographic_df)

display(ppg_resampled_df.head())
```

```python
# Add original index as a column to ppg_resampled_df2
ppg_resampled_df['original_index'] = ppg_resampled_df.index

# Select demographic columns (assuming they are not PPG columns)
demographic_cols = [col for col in ppg_bp_data_online_df.columns if not col.startswith("PPG_")]

# Join demographic data using the original index
ppg_resampled_with_demo = ppg_resampled_df.merge(
    ppg_bp_data_online_df[demographic_cols].reset_index(),
    left_on='original_index',
    right_on='index',
    how='left'
)

# Drop the extra index columns if not needed
ppg_resampled_with_demo = ppg_resampled_with_demo.drop(['original_index', 'index'], axis=1)

display(ppg_resampled_with_demo.head())
```

```python
from scipy.signal import resample

target_length = 200

# Get all PPG columns in the DataFrame
ppg_cols = [col for col in ppg_resampled_with_demo.columns if col.startswith('PPG_')]

# Function to resample a row's PPG columns to target_length
def resample_ppg_row(row, cols, target_len=200):
    arr = row[cols].to_numpy(dtype=float)
    arr = arr[~np.isnan(arr)]
    if len(arr) == 0:
        return np.full(target_len, np.nan)
    return resample(arr, target_len)

# Apply resampling to each row
ppg_resampled_array = np.vstack(
    ppg_resampled_with_demo.apply(lambda row: resample_ppg_row(row, ppg_cols, target_length), axis=1)
)

# Create a new DataFrame with resampled PPG columns
ppg_resampled_df = pd.DataFrame(ppg_resampled_array, columns=[f"ppg_{i}" for i in range(target_length)])

display(ppg_resampled_df.head())
```

```python
# --- STEP 4: Function to resample waveform to fixed length ---
def resample_to_fixed_length(row, cols, target_len=200):
    arr = row[cols].to_numpy(dtype=float)
    # Remove NaNs before resampling (shorten series if needed)
    arr = arr[~np.isnan(arr)]
    if len(arr) == 0:
        return np.full(target_len, np.nan)  # empty signal
    return resample(arr, target_len)
```
